## **Phase I: Synthetic Synthesis (The Lab)**
We utilize **Gemini 2.5 Flash** to programmatically generate "High-Potency Sarcasm." This allows us to stress-test the SOMA Engine against weaponized politeness and nuanced toxicity that traditional datasets often lack.

## **Phase II: Real-World Ingestion (The Wild)**
To prevent **Model Collapse**, we ingest verified real-world datasets (e.g., Jigsaw, DynaHate). This validates SOMA’s performance against actual human linguistic patterns and ensures the engine remains grounded in real-world nuances.

> **Status:** `Active`  
> **Infrastructure:** `Vertex AI Workbench`

```python
# --- STEP 1: INITIALIZE Synthetic FACTORY BRAIN ---
from google import genai
from google.genai import types
import json
from datetime import datetime
import pandas as pd
import json
from datetime import datetime

PROJECT_ID = "project-b75ad945-ea4d-46ac-977"
LOCATION = "us-central1"

client = genai.Client(
    vertexai=True, 
    project=PROJECT_ID, 
    location=LOCATION
)

# Using Gemini 2.5 Flash for rapid synthesis
factory_model = "gemini-2.5-flash" 
# --- SCALED DATA FACTORY LOGIC ---

total_required = 25
samples_per_batch = 5
all_compounds = []

print(f"--- STARTING LARGE-SCALE SYNTHESIS: {total_required} SAMPLES ---")

for i in range(total_required // samples_per_batch):
    batch_prompt = f"""
    Generate a JSON list of {samples_per_batch} unique 'Gray Zone' comments. 
    Focus: Passive-aggressive corporate speak and weaponized politeness.
    Mix: Include at least one completely harmless 'Inert' comment per batch.
    Output format: ["string1", "string2", ...]
    """
    
    response = client.models.generate_content(
        model=factory_model,
        contents=batch_prompt,
        config=types.GenerateContentConfig(response_mime_type="application/json")
    )
    
    batch_list = json.loads(response.text)
    all_compounds.extend(batch_list)
    print(f"Batch {i+1} synthesized. Total: {len(all_compounds)}")

# Now use 'all_compounds' in your Export script
```

```python
# After synthesis loop
print(f"Total compounds after Phase I: {len(all_compounds)}")
```

## **Phase II: The Real-World Pulse**
> **Focus:** `Organic Human Complexity`

To make sure SOMA actually understands how people talk, we’re pulling in real-world data. Synthetic data is great for testing specific theories, but real human language is messy—it has typos, weird slang, and layers of intent that a model can't always "invent." 

This phase is about grounding the AI in the actual "noise" of human interaction to see if it can still find the signal.

*   **Source:** Real-world social datasets.
*   **Goal:** Test SOMA’s ability to decode intent in messy, unpolished text.
*   **The Stress Test:** Can the engine distinguish between "venting" and actual "toxic intent"?

```python
# Use the stable link for hate-speech-research
dataset_url = "https://raw.githubusercontent.com/t-davidson/hate-speech-and-offensive-language/master/data/labeled_data.csv"

def ingest_real_world_data(target_list):
    """
    Ingests samples from a public dataset and appends them to the master list.
    """
    try:
        df = pd.read_csv(dataset_url)
        # Sample 25 entries
        samples = df.sample(25)
        
        # Standardize the text column
        column = 'tweet' if 'tweet' in df.columns else 'comment_text'
        
        for text in samples[column]:
            target_list.append({
                "raw_text": text.strip(),
                "origin": "real_world_dataset",
                "batch_id": "batch_01"
            })
        print(f"Success: Added 25 real-world samples. Total now: {len(target_list)}")
    except Exception as e:
        print(f"Error: {e}")

# Run the function
ingest_real_world_data(all_compounds)
```

# SOMA: Registry & Export (Module 2)
### *Securing the Cognitive Compounds*

---

## **Overview**
This module acts as the "Memory Bank" for the project. It takes the raw input we gathered—both the synthetic edge cases from the lab and the messy reality of the wild—and locks them into a standardized format. 

By using **JSONL**, we ensure that each comment is treated as an individual "thought" or record, making it easy for the SOMA Engine to process them one by one without getting overwhelmed by the volume.

## **The Hand-Off**
*   **Context Preservation:** We keep the 'Origin' tag so the engine knows if it's looking at an AI-generated scenario or a real human interaction.
*   **Scalability:** This format is designed to grow. Whether we are testing 50 samples or 5,000, the hand-off to the main engine remains seamless.
*   **Traceability:** Every batch is timestamped, creating a clear history of how SOMA’s understanding of "Gray Zone" content evolves over time.

> **Status:** `Registry Locked`  
> **Format:** `JSON Lines (.jsonl)`  
> **Next Step:** `Ingestion for Cognitive Adjudication`

```python
# --- STEP 3: EXPORT TO SOMA TEST REGISTRY ---

# Use the stable link for hate-speech-research
dataset_url = "https://raw.githubusercontent.com/t-davidson/hate-speech-and-offensive-language/master/data/labeled_data.csv"

def ingest_real_world_data(target_list):
    """
    Ingests samples from a public dataset and appends them to the master list.
    """
    try:
        df = pd.read_csv(dataset_url)
        # Sample 25 entries
        samples = df.sample(25)
        
        # Standardize the text column
        column = 'tweet' if 'tweet' in df.columns else 'comment_text'
        
        for text in samples[column]:
            target_list.append({
                "raw_text": text.strip(),
                "origin": "real_world_dataset",
                "batch_id": "batch_01"
            })
        print(f"Success: Added 25 real-world samples. Total now: {len(target_list)}")
    except Exception as e:
        print(f"Error: {e}")

# Run the function
ingest_real_world_data(all_compounds)
```

```python
def export_to_registry(data_list):
    """
    Serializes the current list of compounds to a JSONL file.
    """
    timestamp = datetime.now().strftime("%Y%m%d_%H%M")
    filename = f"soma_registry_batch_{timestamp}.jsonl"
    
    with open(filename, 'w') as f:
        for entry in data_list:
            f.write(json.dumps(entry) + '\n')
            
    print(f"Registry created: {filename} with {len(data_list)} compounds.")

# Final Export
export_to_registry(all_compounds)
```
